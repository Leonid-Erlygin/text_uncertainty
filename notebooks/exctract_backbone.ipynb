```python
model_name = "topic_yahoo"  # "topic_agnews"  # "topic_dbpedia" #"topic_yahoo"
out_dir = Path(f"/app/model_weights/backbone/{model_name}")
out_dir.mkdir(exist_ok=True, parents=True)

# 1. Load checkpoint
ckpt_path = f"/app/outputs/{model_name}/last.ckpt"
checkpoint = torch.load(ckpt_path, map_location="cpu")

ckpt = torch.load(ckpt_path, map_location="cpu")

# Extract inner iresnet state dict (remove "backbone.backbone." prefix)
bert_embedder_state = {}
for k, v in ckpt["state_dict"].items():
    if k.startswith("backbone."):
        if "backbone.backbone" in k:
            bert_embedder_state[k.replace("backbone.backbone", "backbone")] = v
        elif "backbone." in k:
            bert_embedder_state[k.replace("backbone.", "")] = v
torch.save(bert_embedder_state, out_dir / "backbone.pth")

weight = ckpt["state_dict"]["softmax_weights"]

# Save as raw tensor (not a dict!)
torch.save(weight, out_dir / "softmax_weight.pt")

print(f"✅ Saved weights: {weight.shape}")
```
